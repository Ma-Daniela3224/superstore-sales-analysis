# Insights Finales - Analisis de Ventas Superstore

## Resumen ejecutivo
El presente análisis se realizó sobre un conjunto de 9,994 transacciones de ventas de una tienda minorista (Superstore), con el objetivo de identificar los factores que afectan la rentabilidad del negocio y descubrir oportunidades de mejora mediante análisis exploratorio de datos (EDA).

Los resultados muestran que la principal causa de las pérdidas no es el volumen de ventas, sino la política de descuentos, especialmente cuando éstos superan el 30%, afectando principalmente determinadas categorías, productos y estados.

## Insight 1: Los descuentos superiores al 30% representan el principal punto de quiebre de la rentabilidad
El análisis evidenció una relación negativa entre el descuento aplicado y la ganancia obtenida. Aunque la correlación entre ambas variables es moderada, el comportamiento cambia de forma importante cuando el descuento supera el 30%.

Hallazgos
- El 18.7% de todas las transacciones generan pérdidas.
- Las transacciones con pérdida presentan un descuento promedio del 48.09%, mientras que las transacciones rentables tienen un descuento promedio de únicamente 8.14%.
- A partir del 30% de descuento, la ganancia promedio por operación se vuelve fuertemente negativa.
- Los descuentos extremos (>40%) concentran las pérdidas individuales más severas del conjunto de datos.

Conclusión

Más que un problema de ventas, el análisis demuestra que la política de descuentos es el principal factor que deteriora la rentabilidad del negocio.

## Insight 2: La categoría Furniture concentra la mayor parte del problema de rentabilidad
Aunque Furniture representa una parte importante de las ventas, su margen de ganancia es considerablemente inferior al de las demás categorías.

Al profundizar el análisis por subcategoría se encontró que el problema no afecta a todos los productos por igual.

Hallazgos
- Furniture registra el margen de ganancia más bajo del negocio (2.49%).
- La subcategoría Tables presenta el peor desempeño, con un margen aproximado de −8.56%.
- Tables también posee uno de los descuentos promedio más altos dentro de Furniture (≈26%).

Conclusión

Las pérdidas de Furniture están altamente concentradas en la subcategoría Tables, por lo que representa un punto prioritario para revisar precios, promociones y estrategias comerciales.

## Insight 3: El mayor volumen de ventas no garantiza mayor rentabilidad
El análisis temporal permitió identificar un comportamiento estacional claro durante el año.

Hallazgos
- Noviembre registra el mayor volumen de ventas del periodo analizado.
- Diciembre y septiembre también presentan niveles elevados de ventas.
- Sin embargo, noviembre muestra uno de los márgenes de ganancia más bajos del año.

Conclusión

El incremento de ventas durante temporadas de alta demanda parece estar acompañado de promociones o descuentos agresivos que reducen significativamente el margen de utilidad.

Esto demuestra que incrementar las ventas no necesariamente implica incrementar la rentabilidad.

## Insight 4: Texas, Illinois y Ohio presentan problemas estructurales de rentabilidad
El análisis geográfico permitió evaluar el margen de ganancia por estado considerando también el tamaño de muestra para evitar conclusiones basadas en pocos registros.

Hallazgos

Los estados con pérdidas más consistentes son:

|Estado |	Margen	| Transacciones |
|---|---|---|
|Ohio |	-21.69% |	469|
|Illinois |	-15.73%	| 492|
|Texas |	-15.12%	| 985|

Aunque algunos estados presentan márgenes muy altos (como Maine o District of Columbia), éstos cuentan con muy pocas transacciones, por lo que sus resultados no son estadísticamente representativos.

Conclusión

Ohio, Illinois y Texas representan problemas de rentabilidad confirmados debido a su combinación de márgenes negativos y un elevado volumen de operaciones, lo que los convierte en prioridades para futuras investigaciones comerciales.

## Insight 5: Existen productos específicos que generan pérdidas desproporcionadas
Además de las tendencias generales, el análisis identificó productos individuales que concentran pérdidas significativas.

Hallazgos
- Algunos productos tecnológicos y mesas de conferencia generan pérdidas acumuladas muy superiores al promedio.
- Estas pérdidas suelen estar asociadas a pocas transacciones con descuentos muy elevados.
- También se identificó que el caso del cliente Sean Miller corresponde a una anomalía puntual y no representa un patrón general del negocio.

Conclusión

El monitoreo de productos con pérdidas recurrentes puede permitir acciones específicas como ajustes de precio, eliminación del catálogo o revisión de proveedores.

## Recomendaciones de negocio
A partir de los resultados obtenidos se proponen las siguientes acciones:

- Revisar la política de descuentos, estableciendo un límite cercano al 30%, ya que a partir de este punto la rentabilidad disminuye de forma considerable.
- Analizar la estrategia comercial de la subcategoría Tables, revisando costos, precios y promociones para reducir las pérdidas concentradas en este segmento.
- Evaluar las campañas promocionales de noviembre, buscando mantener el alto volumen de ventas sin comprometer el margen de ganancia mediante  descuentos excesivos.
- Realizar un análisis específico para Texas, Illinois y Ohio, identificando si las pérdidas se relacionan con determinados clientes, productos, vendedores o políticas comerciales regionales.
- Implementar un monitoreo continuo de productos con pérdidas recurrentes, permitiendo detectar oportunamente artículos que afecten la rentabilidad.

## Limitaciones
Este estudio presenta algunas limitaciones que deben considerarse al interpretar los resultados:

- El conjunto de datos únicamente incluye información histórica de ventas; no incorpora costos operativos, inventario ni información logística.
- El análisis identifica relaciones entre variables, pero no demuestra causalidad absoluta.
- Algunos estados presentan muy pocas transacciones, por lo que sus márgenes pueden no ser representativos.
- Sería recomendable complementar el estudio con información sobre campañas comerciales, costos de adquisición, proveedores y comportamiento de clientes para validar las causas identificadas.
